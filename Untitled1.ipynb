```python
import pandas as pd

orders = pd.read_csv("/content/olist_orders_dataset.csv")
items = pd.read_csv("/content/olist_order_items_dataset.csv")
products = pd.read_csv("/content/olist_products_dataset.csv")

print("All datasets loaded successfully")
```

```python
print("ORDERS DATA")
print(orders.head())

print("\nITEMS DATA")
print(items.head())

print("\nPRODUCTS DATA")
print(products.head())
```

```python
print("ORDERS DATA")
print(orders.head())

print("\nITEMS DATA")
print(items.head())

print("\nPRODUCTS DATA")
print(products.head())
```

```python
items["revenue"] = items["price"]

total_revenue = items["revenue"].sum()

print("TOTAL REVENUE:")
print(total_revenue)
```

```python
# merge items with products
merged = items.merge(products, on="product_id")

# top categories by revenue
top_categories = merged.groupby("product_category_name")["price"].sum().sort_values(ascending=False).head(10)

print("TOP 10 PRODUCT CATEGORIES BY REVENUE")
print(top_categories)
```

```python
import matplotlib.pyplot as plt

top_categories = merged.groupby("product_category_name")["price"].sum().sort_values(ascending=False).head(10)

plt.figure(figsize=(10,5))
top_categories.plot(kind="bar")

plt.title("Top 10 Product Categories by Revenue")
plt.xlabel("Product Category")
plt.ylabel("Revenue")

plt.xticks(rotation=45)
plt.show()
```
